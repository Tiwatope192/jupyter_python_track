```python
def load_to_snowflake(conn, dfs_dict, database, schema):
    for table_name, df in dfs_dict.items():

        success, nchunks, nrows, _ = write_pandas(
            conn,
            df,
            table_name,
            auto_create_table=True,
            database='DATABASE3',
            schema='BRONZE',
        )
        
```
